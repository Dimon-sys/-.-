### Замена строк с помощью метода `sub()`

Объекты Regex имеют метод `sub()`, который принимает два аргумента. 

Первый аргумент — это строка, которая должна подставляться вместо найденных совпадений. 

Второй аргумент – это строка регулярного выражения. 

Метод `sub()` возвращает строку, к которой применены замены.


```python
import re
```

```python
name = re.compile(r'Агент(у)? \w+')
name.sub('****', 'Агент Алиса передала секретные документы Агенту Ивану')
```

```python
name = re.compile(r'\d+')
name.sub('Скрыто', 'Мой пароль: 2355753465764')
```

Иногда возникают ситуации, когда совпавший с регулярным выражением текст сам используется в качестве части подстановочного текста. 

Для этого можно использовать в первом аргументе при вызове метода `sub()` обратные ссылки `\1, \2, \3` и т.д., 
которые имеют следующий смысл: 
"Вставить в подстановочный текст группы 1, 2, 3 и так далее".


Предположим, что мы хотим скрыть имена секретных агентов, отображая в них лишь первые буквы. 

Для этого можно воспользоваться регулярным выражением `Агент(\w)\w* `
и передать методу `sub()` в качестве первого аргумента строку `r'\1****'`. 

Ссылка `\1` в этой строке будет заменяться тем текстом, который будет захвачен группой 1, т.е. группой `(\w)`регулярного выражения.


```python
name = re.compile(r'Агент(у)? (\w)\w*',re.I)
name.sub(r'\2****', 'Агент Алиса сказала агенту Катя, что агент Анна знает, что агент Боб был двойным агентом')
```
