```python
import openpyxl
import json
```

```python
wb = openpyxl.load_workbook('example2.xlsx')
```

```python
sheet = wb.active
```

```python
data = {}
```

```python
for row in sheet.iter_rows(min_row=1,
                           max_row=8, 
                           min_col=1,
                           max_col=5,
                           values_only=True):
    print(row)
```

```python
for row in sheet.iter_rows(min_row=2,
                           max_row=8, 
                           min_col=1,
                           max_col=5,
                           values_only=True):
    data_id = row[0]
    d = {'name': row[1],
         'sum': row[2] * row[3]
         }
    data[data_id] = d
```

```python
print(data)
```

```python
file_data = open('data.json',"w")
```

```python
json.dump(data, file_data)
```

```python
file_data.close()
```

```python
wb.save('example2.xlsx')
```
