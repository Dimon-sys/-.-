```python
import openpyxl
import random
```

```python
def func(x):
    return x ** 2 - 5 * x + 10
```

```python
wb = openpyxl.Workbook()
```

```python
sheet = wb.active
```

```python
n, m = 10, 2
```

```python
for i in range(10):
    x = random.randint(1,50)
    row = [x, func(x)]
    sheet.append(row)
```

```python
for row in sheet.iter_rows(values_only=True):
    print(row)
```

```python
wb.save('example1.xlsx')
```
