```python
import openpyxl
```

```python
wb = openpyxl.load_workbook('example.xlsx')
```

```python
sheet = wb.active
```

```python
n = sheet.max_row
print(n)
```

```python
m = sheet.max_column
print(m)
```

```python
for row in sheet.iter_rows(min_row=1,
                           max_row=n, 
                           min_col=1,
                           max_col=m):
    print(row)
```

```python
for row in sheet.iter_rows(min_row=1,
                           max_row=n, 
                           min_col=1,
                           max_col=m,
                           values_only=True):
    print(row)
```

```python
for column in sheet.iter_cols(min_row=1,
                           max_row=n, 
                           min_col=1,
                           max_col=m):
    print(column)
```

```python
for column in sheet.iter_cols(min_row=1,
                           max_row=n, 
                           min_col=1,
                           max_col=m,
                           values_only=True):
    print(column)
```

```python
for row in sheet.rows:
    print(row)
```

```python
for row in sheet.iter_rows(values_only=True):
    print(row)
```

```python
wb.save('example.xlsx')
```
