#### Полная версия программы

```python
import zipfile
import os

def to_zip(folder):
    folder = os.path.abspath(folder)
    number = 1
    while True:
        zip_name = f'{os.path.basename(folder)}_{number}.zip'
        if not os.path.exists(zip_name):
            break
        number += 1
    
    print(f'Создание файла {zip_name}')
    backup = zipfile.ZipFile(zip_name,'w')
    
    for foldername, subfolders, filenames in os.walk(folder):
        print(f'Добавляем файлы из папки {foldername}')
        backup.write(foldername)
        for filename in filenames:
            if filename.startswith(os.path.basename(folder)) and filename.endswith('.zip'):
                continue
            backup.write(os.path.join(foldername, filename))
    backup.close()


to_zip('C:\\Python')
print('DONE!')
```
