# Кортежи

#### **Кортежи (tuple)** — это те же списки, только неизменяемые. Над ними можно производить те же операции, что и над списками, — кроме тех, которые изменяют кортеж.

```python
t = (1, 5, 2, 6)
t[0] = 100
```

```python
lst = [1,6, 8, 9]
t = tuple(lst)
print(t)
```

```python
t = (1)
print(t, type(t))
```

```python
t += 2
print(t)
```

```python
t = (1,)
print(t, type(t))
```

```python
t1 = (1,)
t2 = (4, 7, 3, 2)
t = t1 + t2
print(t)
```

```python
m = [[1, 2, 3], [4, 5, 6]]
print(m)
```

```python
for i in range(len(m)):
    for j in range(len(m[0])):
        print(m[i][j],end=" ")
    print()
```

```python
m = [[0,0,0] for i in range(4)]
print(m)
```

```python
b = m
print(b)
```

```python
m[0][0] = 1000
print(b)
print(m)
```

```python
m.append(500)
print(m)
print(b)
```

```python
a = [0, 0, 0]
m = [a for _ in range(2)]
print(m)
```

```python
m[0][0] = 1000
print(m)
```

```python
a = [1, 2, 3, 4, 5, 6]
for i in range(len(a)-1):
    if a[i] == a[i+1]:
        print('y')

a = [1, 2, 3, 4, 5, 6]
for x in a:
    x = 10
    print(x)
print(a)

```

```python
a = (3, 5, [1, 0, 1])
a[2][0]= 1000
print(a)
```
