# Модуль pickle

`pickle` (англ. консервировать, мариновать) — модуль сериализации и десериализации объектов в Python для последующей их передачи.

pickle умеет запаковывать:
- None, True, False
- строки (обычные или Юникод)
- стандартные числовые типы данных
- словари, списки, кортежи
- функции
- классы

Импортируется модуль стандартно:

```python
import pickle
```

Модуль  `pickle` имеет два метода:
- `dump`, который записывает  данные в файл
- `load`, который загружает данные из файла.


## Запись данных в бинарный файл

Модуль `pickle` предоставляет следующую функцию:

`dump(<Объект>, <Файл>) `

производит сериализацию объекта и записывает данные в указанный файл.  

В параметре `<Файл>` указывается файловый объект, открытый для записи.

`pickle.dump(obj, file, protocol=None, *, fix_imports=True)`

`protokol` - число, формат выводимых данных. Если в аргументе `protocol` передается отрицательное значение, выбирается наиболее современный протокол. 

0 - текстовый формат, обратно совместимый с более ранними версиями Python.

1 – двоичный формат, который также совместим с большинством предыдущих версий Python.

2 – новый формат, который обеспечивает наиболее эффективный способ сохранения классов и экземпляров.

3 – используется в Python 3 и несовместим с предыдущими версиями интерпретатора.

4 - добавлен Python 3.4, добавлена поддержка очень больших объектов. 

Самая последняя версия протокола хранится в переменной `pickle.HIGHEST_PROTOCOL`. 

Если объект не поддерживает возможность сериализации, возбуждается исключение `pickle.PicklingError`.



### Пример 1.

```python
import pickle

myfile = open('file.data', 'wb')
obj = [2, 'hello', '56756757', (2, 6, 1, 10000), [100, 222, 666], {1:20, 'a':23}]
pickle.dump(obj, myfile)
myfile.close()
```

### Чтение данных из бинарного файла

Чтобы прочитать данные из бинарного файла, используется метод `load()`

`load(<Файл>)`

читает сохраненные данные из файла и преобразует их в объект. 

В параметре `<Файл>` указывается файловой объект, открытый для чтения. 

```python
import pickle

myfile = open('file.data', 'rb')
x = pickle.load(myfile)
print(x, x[0], type(x[-1]))
myfile.close()
```

Данные читаются в том же порядке, в котором они были записаны. 

```python
import pickle

myfile = open('data.bin', 'wb')
a1 = [1, 100, 333, 8990]
a2 = [('Иванов', 'Иван'), ('Петров', 'Петр'), ('Петрова','Елизавета')]
a3 = ['127.0.0.1', '192.255.200.193']
pickle.dump(a1, myfile)
pickle.dump(a2, myfile)
pickle.dump(a3, myfile)
myfile.close()
```

```python
import pickle

myfile = open('data.bin', 'rb')
x1 = pickle.load(myfile)
x2 = pickle.load(myfile)
x3 = pickle.load(myfile)
#x4 = pickle.load(myfile)
print(x1)
print(x2)
print(x3[0])
myfile.close()
```

По достижении конца файла возбуждается исключение `EOFError`.

```python
import pickle

myfile = open('data.bin', 'rb')
for x in myfile:
    x1 = pickle.load(myfile)
    print(x1)
myfile.close()
```

Если файл содержит поврежденные данные, которые не могут быть декодированы, функция возбуждает исключение `pickle.UnpicklingError`.
