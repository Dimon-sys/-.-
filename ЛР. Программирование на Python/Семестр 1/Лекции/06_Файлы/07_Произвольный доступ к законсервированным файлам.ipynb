# Произвольный доступ к законсервированным файлам

Представим, что мы законсервировали несколько списков, а теперь хотим получить доступ к одному произвольному cписку. 

То есть мы не хотим читать все списки подряд, пока не прочитаем тот, который нам нужен.


Для этого есть модуль `shelve`.

Первым делом создадим полку `s`.

Данная функция работает подобно файловой функции `open()`, но открывает не текстовый файл, а файл с консервированными объектами. 

Доступ к полке мы будем осуществлять через переменную `s`, которая работает как словарь, содержимое которого хранится в файле.

```python
import shelve

s = shelve.open('names.dat')
```

Помещаем данные на полку:

```python
s['first_names'] = ['Петр', 'Иван']
s['last_names'] = ['Петров', 'Иванов']
s['age'] = [18, 25]
s['1'] = 1998
s.sync()
s.close()
```

Выполнив этот код на Windows-компьютере, вы увидите в текущем рабочем каталоге три новых файла: `names.bak`, `names.dat` и `names.dir`.
На компьютере, работающем под управлением OS X, будет создан только один файл `names.db`.



Описанные двоичные файлы содержат данные, которые вы сохранили в хранилище. 

Точный формат хранения данных в этих двоичных файлах для нас не имеет значения; нам достаточно знать лишь то, что именно делает модуль shelve, а не как он это делает

Теперь откроем файл полки и прочитаем просто список фамилий:

```python
import shelve

a = shelve.open('names.dat')
print(a)
print(a['last_names'])
```

Как и словари, хранилища имеют методы `keys()` и `values()`, извлекающие из хранилища коллекции ключей и значений, подобные спискам. 

Поскольку возвращаемые этими методами коллекции лишь подобны спискам, а не являются истинными списками, для того чтобы получать их в виде списков, их следует передавать функции list(). 


```python
print(a.keys())
print(list(a.keys()))
```

```python
print(a.values())
print(list(a.values()))
```
