# Хранение структурированных данных в файлах

Текстовые файлы хороши тем, что их можно просматривать и редактировать в любом текстовом редакторе. 
Но функциональность текстовых файлов ограничена. 
В определенный момент любому Python-программисту понадобится запаковать какой-то объект и спрятать его до лучших времен. 

Скажем, в `config-файл`. Как это сделать?

Мы не можем взять и сохранить в них любые данные, например, словарь или список. 
Придется изобретать какой-то формат файла или использовать уже существующие, например, `XML`.

Но Python позволяет хранить в файлах структурированные данные.

В Python есть понятие **консервации**.

**Консервация** – это долговременное хранение структурированных данных.

**Сериализация** – процесс перевода какой-либо структуры данных в последовательность битов. 

Восстановление начального состояния структуры данных из битовой последовательности называется **десериализацией** (структуризацией).
 
Эта операция является обратной к операции сериализации.


Практически любой тип объекта может быть сохранен на диске в любой момент его жизни, а позже прочитан с диска. 
Сохранить объекты в файл и в дальнейшем восстановить объекты из файла позволяют модули `pickle` и `shelve`. 

Модуль `pickle` позволяет консервировать структуры данных и сохранять их в файлах, а модуль `shelve` обеспечивает произвольный доступ к законсервированным объектам.

Для записи бинарных (двоичных) данных нам нужно открыть файл в двоичном режиме.

В `Python 2` имеется модуль `cPickle`, содержащий реализацию функций модуля `pickle` на языке `C`. 
Он обладает гораздо более высокой производительностью по сравнению с модулем `pickle`, но ограничивает возможность создания производных классов от `Pickler` и `Unpickler`. 

В `Python 3` также имеется модуль, содержащий реализацию на языке `C`, но он не используется непосредственно (модуль `pickle` использует его автоматически).
